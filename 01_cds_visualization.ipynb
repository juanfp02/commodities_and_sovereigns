# CDS Visualization

This notebook reads the CDS sheet from the masterfile and plots the sovereign CDS spreads.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read CDS sheet from masterfile
df = pd.read_excel('data/raw/data_masterfile.xlsx', sheet_name='CDS')
df.set_index('Dates', inplace=True)
df.head()
```

```python
# Clean column names (remove _USD suffix)
df.columns = [col.replace('_USD', '') for col in df.columns]
df.columns
```

```python
# Plot all CDS spreads
fig, ax = plt.subplots(figsize=(14, 8))

for col in df.columns:
    ax.plot(df.index, df[col], label=col, alpha=0.8)

ax.set_xlabel('Date')
ax.set_ylabel('CDS Spread (bps)')
ax.set_title('Sovereign CDS Spreads')
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
