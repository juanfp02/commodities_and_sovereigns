```python
import yfinance as yf
import pandas as pd

# Get Brent crude and OVX
brent = yf.download('BZ=F', start='2008-01-01')
ovx = yf.download('^OVX', start='2008-01-01')

# Check what you got
print(brent.columns)
print(brent.head())

# Extract Close - handle multi-level columns
if isinstance(brent.columns, pd.MultiIndex):
    brent_close = brent['Close']['BZ=F']
    ovx_close = ovx['Close']['^OVX']
else:
    brent_close = brent['Close']
    ovx_close = ovx['Close']

# Resample to weekly
brent_weekly = brent_close.resample('W-FRI').last()
ovx_weekly = ovx_close.resample('W-FRI').last()

# Check they're Series
print(type(brent_weekly))
print(brent_weekly.head())

# Now combine
weekly_data = pd.concat([brent_weekly, ovx_weekly], axis=1)
weekly_data.columns = ['Brent', 'OVX']
weekly_data = weekly_data.reset_index()
weekly_data.columns = ['Date', 'Brent', 'OVX']

print(weekly_data.head())
```

```python
weekly_data.to_csv('brent_ovx_weekly.csv', index=False)
```
