```python
import pandas as pd
import pycountry

# Read the GDP CSV file
df = pd.read_csv('weo_gdp_ngdpd_annual_2000_2024.csv')

# Function to convert ISO alpha-3 code to country name
def iso_to_country_name(iso_code):
    try:
        return pycountry.countries.get(alpha_3=iso_code).name
    except (AttributeError, KeyError):
        return iso_code  # Return original code if not found

# Transform the country column
df['country'] = df['country'].apply(iso_to_country_name)

# Display the first few rows to verify
print(df.head(10))
```

```python
df.to_csv('weo_gdp_ngdpd_annual_2000_2024_transformed.csv', index=False)
```
