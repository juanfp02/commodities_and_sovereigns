```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('More_CDS_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
```

```python
df
```

```python
# Plot all CDS spreads
fig, ax = plt.subplots(figsize=(14, 8))

for col in df.columns:
    ax.plot(df.index, df[col], label=col, alpha=0.8)

ax.set_xlabel('Date')
ax.set_ylabel('CDS Spread (bps)')
ax.set_title('Sovereign CDS Spreads')
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
CDS_df = pd.read_excel('data_masterfile.xlsx', sheet_name='CDS')
CDS_df = CDS_df.rename(columns={'Dates': 'Date'})
CDS_df['Date'] = pd.to_datetime(CDS_df['Date'])
CDS_df.set_index('Date', inplace=True)
CDS_df.columns = [col.replace('_USD', '') for col in CDS_df.columns]
```

```python
merged = pd.merge(df, CDS_df[['Colombia','Morocco']], on='Date', how='inner')
```

```python
merged.to_csv('use_this_cds_data.csv', index=True)
```
